```python
plt.figure(figsize=(8,5))
sns.histplot(df['Price (Euro)'], bins=20)
plt.title("Distribution of Laptop Prices")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Weight (kg)'], bins=20)
plt.title("Distribution of Laptop Weight")
plt.xlabel("Weight")
plt.ylabel("Count")
plt.show()
```

```python
company_revenue = df.groupby('Company')['Price (Euro)'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
company_revenue.plot(kind='bar')
plt.title("Total Revenue by Company")
plt.ylabel("Total Price")
plt.xlabel("Company")
plt.show()
```

```python
top_products = df.groupby('Product')['Price (Euro)'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_products.plot(kind='bar')
plt.title("Top 10 Products by Total Revenue")
plt.ylabel("Total Price")
plt.xlabel("Product")
plt.xticks(rotation=45)
plt.show()
top_products
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Inches', y='Price (Euro)', data=df)
plt.title("Screen Size vs Price")
plt.xlabel("Inches")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['Inches', 'Price (Euro)']].corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```
